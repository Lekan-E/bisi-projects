# Linear Regression with Multiple Features

### Using 2 features predict `salary`

```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
# import the model
from sklearn.linear_model import LinearRegression
# import RMSE
from sklearn.metrics import root_mean_squared_error

#import the dataset
df = pd.read_csv('Pay.csv')
df.shape
```

```python
# find correlations
df.corr()
```

### Feature Selection


```python

plt.subplot(2,2,1)
plt.scatter(df.Age, df.Salary)
plt.xlabel('Age')
plt.ylabel('salary')
plt.title('Age vs Salary')


plt.subplot(2,2,2)
plt.scatter(df.Experience, df.Salary)
plt.xlabel('Experience')
plt.ylabel('salary')
plt.title('Experience vs Salary')
plt.show()
```

## Now let's build a simple linear regression model

We will use the variable `Age` to predict `Salary`

```python
df.head()
```

```python
# Seperate features and target
x = df[['Age']]
y = df[['Salary']]
```

```python
plt.scatter(x,y,color='lightgray')
plt.xlabel('Age')
plt.ylabel('salary')
plt.title('Age vs Salary')
plt.show()
```

```python
# Create an instance of the model
Regression = LinearRegression()
```

```python
# train the model
model = Regression.fit(x,y)
```

```python
# get the parameters
print('Model coefficiant is M', model.coef_)
print('Model intercept is C', model.intercept_)
print('Y = ', model.coef_.item(), 'x + ', model.intercept_.item())
```

```python
# Predictions with new model
y_pred = model.predict(x)
```

```python
plt.scatter(x,y,color='lightgray')
plt.plot(x, y_pred, color='red', linewidth=2, label='Regression line')
plt.xlabel('Age')
plt.ylabel('salary')
plt.title('Age vs Salary')
plt.show()
```

```python
rmse1 = root_mean_squared_error(y_pred, y)
rmse1
```
